## 2. Lists are mutable

**Mutable** means a list can be changed in place. Assigning to an element changes
the list object. This differs from assigning a name to a new object.

```python
scores = [10, 20, 30]
scores[0] = 15
print(scores)
```

### Replace a slice instead of one item

The slice selects two items. Assignment replaces them with the two supplied items
and changes the same list in place. Reading a slice and assigning to a slice are different operations.

```python
fam = ["liz", 1.73, "emma", 1.68, "mom", 1.71, "dad", 1.89]
fam[1:3] = [1.8, "jenny"]
print(fam)
```

### Another name is an alias, not a copy

`alias = scores` makes both names refer to the same list. A change through either name
is visible through the other. `is` tests whether two names refer to the same object.

```python
scores = [10, 20, 30]
alias = scores
alias[0] = 99
print("scores:", scores)
print("alias: ", alias)
print(alias is scores)  # Both names refer to the same list.
```

### `append` adds one item; `extend` adds each supplied item

A list can itself be an item. Appending a list adds it as one nested item;
extending with a list adds its items individually. Both methods change the original list.

```python
values = [1, 2]
values.append([3, 4])
print(values)
print(len(values))  # Three items: two integers and one list.
```

```python
values = [1, 2]
values.extend([3, 4])
print(values)
print(len(values))  # Four items, all integers.
```

Extend also adds the supplied items individually when the items have different types.

```python
fam = ["liz", 1.73, "emma", 1.68, "mom", 1.71, "dad", 1.89]
fam.extend(["joe", 2.0])  # Two new outer items, added at the end.
print(fam)
```

## 4. What happens when a function receives a list?

The parameter name refers to the same list that the caller supplied. A function can
therefore change that list through the parameter. No `global` statement is needed.

This function intentionally updates its input and has no explicit return statement.

```python
def add_score(score_list, new_score):
    """Add a score to the supplied list in place."""
    score_list.append(new_score)


scores = [10, 20]
result = add_score(scores, 30)
print("caller's list:", scores)
print("return value: ", result)
```

### Return a new list when the input should stay unchanged

Copy the input first, change the copy, and return it. The caller can choose a name
for the result. Compare both the original list and the returned list.

```python
def with_score(score_list, new_score):
    """Return a new list with one additional score."""
    updated_scores = score_list.copy()
    updated_scores.append(new_score)
    return updated_scores


scores = [10, 20]
updated_scores = with_score(scores, 30)
print("new list:", updated_scores)
print("original:", scores)
```

`in` tests whether a value is present. `remove(value)` removes the **first matching value**,

```python
values = [10, 20, 20, 30]
print(20 in values)
print(99 in values)
values.remove(20)  # Remove a value, not index 20.
print(values)  # The second 20 remains.
```

`.pop(index)` removes and **returns** the item at an index. With no index, it uses the last item.
Unlike `append`, it returns the removed item rather than `None`.

```python
scores = [10, 20, 30]
removed_score = scores.pop()
print(removed_score)
print(scores)
```

```python
scores = [10, 20, 30]
removed_score = scores.pop(0)
print(removed_score)
print(scores)
```

### Pop from the end versus the beginning

Keep the removed item under a name, then inspect the remaining list. With items
added at the end, removing from the end gives last-in, first-out order; removing
from the beginning gives first-in, first-out order.

```python
fam = ["liz", 1.73, "emma", 1.68, "mom", 1.71, "dad", 1.89]
j = fam.pop()  # if you don't specify an index, it pops the last item in the list
# default behavior of pop() without any arguments is like a stack. last in first out
print(j)
print(fam)
```

```python
fam = ["liz", 1.73, "emma", 1.68, "mom", 1.71, "dad", 1.89]
j = fam.pop(0)  # you can also specify an index.
# Using index 0 makes pop behave like a queue. first in first out
print(j)
print(fam)
```

## 6. Sorting: change the list or return a new list?

```python
scores = [7, 2, 9, 4]
result = scores.sort()
print(scores)
print(result)
```

## 10. Turn text fields into numeric data

Use the tools in order: split the text, strip each field, convert it with `int`,
and append the number to a new list. This example assumes every field is a valid integer.

The loop version makes each step visible. A compact comprehension version appears
in the optional extension after we understand this version.

```python
raw_scores = " 8, 10, 6 "
fields = raw_scores.split(",")
print(fields)

numeric_scores = []
for field in fields:
    clean_field = field.strip()
    numeric_scores.append(int(clean_field))

print(numeric_scores)
print(type(numeric_scores[0]))
print(repr(raw_scores))  # Processing created new objects; the input is unchanged.
```

The same conversion loop also works when the numeric fields have no surrounding spaces.

```python
num_string = "2,3,4,7,8"
numbers = []
for field in num_string.split(','):
    numbers.append(int(field))
numbers
```

### Why conversion matters

Strings are compared character by character, not as numeric quantities. Convert the
fields before doing numerical calculations or sorting them as numbers.

```python
text_scores = ["10", "2", "30"]
numeric_scores = []
for field in text_scores:
    numeric_scores.append(int(field))

print("text order:   ", sorted(text_scores))
print("numeric order:", sorted(numeric_scores))
print("total:        ", sum(numeric_scores))
```

### Find a position: two ways to report a missing match

Both `find()` and `index()` return the first matching position. `find()` returns `-1`
if the substring is missing; `index()` raises `ValueError` instead.

Position `0` is a valid match. A missing-match result of `-1` is a signal to check,
not a position to use for indexing: indexing with `-1` would select the last character.

```python
fruit = "bananas"
print(fruit.find("b"))
print(fruit.find("na"))
print(fruit.index("na"))
print(fruit.find("x"))
```

An uppercase letter can also be searched for; this returns the first uppercase `A` in the course description.

```python
name.index('A') # index of the first instance
```

**Expected error:** `index()` reports a missing substring with `ValueError`.

```python
fruit.index('x')  # Expected ValueError.
```

Check the result explicitly before using it. `if position:` would be incorrect here:
it treats a valid match at `0` as false and the missing-match value `-1` as true.

```python
fruit = "bananas"
for target in ["b", "x"]:
    position = fruit.find(target)
    if position != -1:
        print(target, "starts at index", position)
    else:
        print(target, "was not found")
```

## Optional: nested lists and shallow copies

A list can contain other lists. Use one index to select a row, then a second index
to select an item in that row. Here each row stores a name and a height in meters.

```python
family = [["Ana", 1.70], ["Bo", 1.80]]
print(family[0])
print(family[0][1])
print(family[-1][-1])
```

A shallow copy makes a new **outer list**, but its items still refer to the same inner lists.
Changing an inner list is therefore visible through both outer lists.
For this example, `copy.deepcopy()` also copies the inner lists.

```python
import copy

family = [["Ana", 1.70], ["Bo", 1.80]]
shallow_family = family.copy()
deep_family = copy.deepcopy(family)
family[0][1] = 1.75

print("original:", family)
print("shallow: ", shallow_family)
print("deep:    ", deep_family)
print(shallow_family is family)        # Different outer lists.
print(shallow_family[0] is family[0])  # Shared inner list.
print(deep_family[0] is family[0])     # Separate inner list.
```

## Optional: combine a loop, append, and simultaneous assignment

Generate positive Fibonacci values up to 100. With `a, b = b, a + b`, Python evaluates
both expressions on the right before assigning either name. Thus the new `b` uses
the previous values of both names. This prepares an idea used again with tuples.

The `break` check happens before `append`, so an over-limit value is not added.

```python
a, b = 0, 1
limit = 100
fibonacci = []
while True:
    a, b = b, a + b
    if a > limit:
        break
    fibonacci.append(a)
print(fibonacci)
```

## Optional: implement a single-character search

Now that we know what `find()` does, implement a simpler version using a loop.
This function assumes `target` is one character. It returns as soon as a match is
found; `-1` is returned only after every character has been checked.

The built-in `find()` also handles multi-character substrings; this function does not.

```python
def find_letter(text, target):
    """Return the first index of a one-character target, or -1 if absent."""
    position = 0
    while position < len(text):
        if text[position] == target:
            return position
        position += 1
    return -1


for target in ["b", "a", "x"]:
    print(target, find_letter("bananas", target), "bananas".find(target))
```

### Search the longer text, one call at a time

`myfind` is the same single-character search idea with the original variable names.
Compare its result with `find` and `index`, then compare the two missing-match behaviors.

```python
def myfind(string, letter):
    index = 0
    while index < len(string):
        if string[index] == letter:
            return index
        index = index + 1
    return -1
```

## Optional: text comparison follows character order

Strings compare lexicographically using Unicode code points: compare the first
differing characters; when one is a prefix of the other, the shorter string comes first.
For these examples, digits come before uppercase letters, which come before lowercase letters.

This default order is case sensitive; it does not apply numeric or language-specific
alphabetical rules. The number conversion example in Section 10 is the practical consequence.

```python
print("A" < "B")
print("Z" < "a")
print("1" < "A")
print("0" < "00")   # The first string is a prefix of the second.
print("11" < "101") # Compare the second characters: "1" versus "0".
print("!" < "@")
print(sorted(["bo", "Ada", "ana"]))
```
