## 1. Values have types

A **value** is a piece of data. Its **type** determines what operations make sense.
The built-in function `type()` tells us a value's type.

| Type | Example | Meaning |
| --- | --- | --- |
| `int` | `2` | An integer |
| `float` | `2.0` | A floating-point number |
| `str` | `"2"` | Text, written inside quotes |
| `bool` | `True` | One of two Boolean values: `True` or `False` |
| `NoneType` | `None` | The type of `None`, used to represent the absence of a value |

Predict the type before running each cell. Pay attention to quotes and decimal points.

```python
type(2)
```

```python
type(2.0)
```

```python
type("2")
```

```python
type(True)
```

```python
type(None)
```

`True`, `False`, and `None` are case-sensitive Python spellings.
Putting quotes around them makes text instead.

```python
type("True")
```

## 2. Give values names

An **assignment** uses a single equals sign, `=`. Python evaluates the expression
on the right, then makes the name on the left refer to the resulting object.

Assignment by itself does not display a result.

```python
points_earned = 18
points_possible = 20
```

In a notebook, a value on the last line of a cell is displayed automatically.
Use `print()` to display values explicitly; commas let you print several values on one line.

```python
points_earned
```

```python
print(points_earned, points_possible)
```

### Choose names that explain the data

Use descriptive names and **snake_case**: lowercase words separated by underscores.
For example, `points_earned` is more informative than `p`.

- Names are case-sensitive: `score` and `Score` are different names.
- For the names we write in this course, use letters, digits, and underscores, and do not start with a digit.
- Do not use spaces or Python keywords such as `for` and `return`.
- Avoid replacing built-in names such as `print`, `type`, and `str`.

Names refer to objects. We do not have to declare a name's type before using it.

### Division produces a float

For the integer and float values we are using, `/` produces a float—even when
the result is a whole number.

```python
proportion_earned = points_earned / points_possible
print(proportion_earned)
print(type(proportion_earned))
```

```python
result = 20 / 5
print(result)
print(type(result))
```

Adding or multiplying an integer and a float also produces a float.

```python
print(3 + 0.5)
print(3 * 0.5)
```

### Full groups and leftovers: `//` and `%`

Suppose we have 23 cookies and put 6 in each box.

- `/` gives the ordinary quotient.
- `//` gives the quotient rounded down. For these nonnegative counts, it tells us how many **full boxes** we can make.
- `%` gives the **remainder**: how many cookies are left over.

Predict both counts before running the cell.

```python
cookies = 23
box_size = 6

full_boxes = cookies // box_size
leftover_cookies = cookies % box_size

print(full_boxes, leftover_cookies)
```

We can check that the full boxes and leftovers account for all 23 cookies.

```python
full_boxes * box_size + leftover_cookies
```

## 4. Reassignment changes what a name refers to

In `score = score + 3`, Python first uses the current value of `score` to evaluate
the right side. It then makes `score` refer to the result.

```python
score = 10
score = score + 3
print(score)
```

### Does another name change too?

Assigning `previous_score = score` makes both names refer to the same integer object
at that moment. It does not make `previous_score` follow future assignments to `score`.

Predict both values. We reset `score` here so the example stands on its own.

```python
score = 10
previous_score = score
score = score + 3

print(score, previous_score)
```

Integers cannot be changed in place. The reassignment gives `score` a new binding;
`previous_score` still refers to the original integer.

We will revisit the distinction between reassignment and changing an object when we study lists.

## 5. Strings are text

Single or double quotes create strings. Choose whichever makes the text easier to write.
Here, double quotes allow the apostrophe inside the sentence.

```python
student_name = 'Maya'
message = "I'm learning Python."
print(student_name)
print(message)
```

The same operator can behave differently for different types:
`+` joins strings, and multiplying a string by an integer repeats it.
The built-in function `len()` counts the characters in these strings.

```python
"Hello, " + student_name
```

```python
"ha" * 3
```

```python
len(student_name)
```

### Text that looks like a number is still text

Compare these two results. Quotes change the type and therefore the meaning of `+`.

```python
"12" + "3"
```

## 6. Comparisons produce Boolean values

Use `==` to ask whether two values are equal. This is different from `=`, which assigns a value to a name.
Use `<` to ask whether the left value is less than the right value.
The result of a comparison is `True` or `False`.

```python
points_earned == points_possible
```

```python
points_earned < points_possible
```

Numeric values of different numeric types can compare equal.
A string containing digits is still a different kind of value.

```python
2 == 2.0
```

```python
"2" == 2
```

## 8. Floating-point calculations can be approximate

Integers represent whole numbers exactly, subject to available memory.
Floats use a finite binary representation; many decimal fractions cannot be represented exactly.

**Predict:** will these two ways of calculating three tenths produce exactly equal values?

```python
a = (1 + 2) / 10
b = 1 / 10 + 2 / 10
print(a, b)
```

```python
a == b
```

The tiny difference is a consequence of finite-precision arithmetic.
It is not a mistake in the mathematical identity.

When approximate numerical equality is what we need, `math.isclose()` compares values
using a tolerance. Its default tolerance is sufficient for this example; an appropriate
tolerance in an analysis depends on the scale and purpose of the comparison.

```python
math.isclose(a, b)
```

Floats typically provide about 15–16 significant decimal digits of precision,
not a fixed number of places after the decimal point.
