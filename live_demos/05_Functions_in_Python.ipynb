## 2. Define a function, then call it

Suppose we want to add two bonus points to a quiz score.

```python
def function_name(parameter):
    # indented function body
    return result
```

- `def` defines a function and gives it a name.
- A **parameter** is a name in the function definition, such as `score` below.
- An **argument** is a value supplied when calling the function, such as `8`.
- The colon and indentation mark the function body.
- `return` ends the call and sends a value back to the caller.

Running the definition creates the function. Its body runs only when we call it.

````python
def add_bonus(score):
    new_score = score + 2
    return new_score
````

````python
add_bonus(8)
````

For this call, the parameter `score` refers to the argument `8`.
The function computes `10` and returns it. The notebook displays the returned value.

````python
quiz_score = 14
add_bonus(quiz_score)
````

### Follow the execution

1. Python reaches a function call and evaluates its arguments.
2. It binds the arguments to the parameters and runs the function body.
3. At `return`, the call finishes and its returned value can be used by the caller.
4. Execution continues after the call.

Predict the order of the printed messages below. The assignment stores the returned value;
it does not print that value.

````python
print("Before the call")
updated_score = add_bonus(8)
print("After the call")
print(updated_score)
````

A returned value can also be used directly in another expression.

````python
add_bonus(8) * 2
````

## 3. Printing and returning do different jobs

`add_bonus()` **returns** a value that other code can use.
This new function performs the same calculation but only **prints** the result.

````python
def print_bonus(score):
    print(score + 2)
````

````python
printed_result = print_bonus(8)
````

We saw `10` on the screen, but what was assigned to `printed_result`?

A function that reaches the end without a `return` statement returns `None`.
A bare `return` also returns `None`. Printing a value does not return that value.

````python
print(printed_result)
````

````python
returned_result = add_bonus(8)
print(returned_result)
````

Use `return` when the caller needs the result for another calculation.
Use `print` when you want to display something. A function can do both,
but these are separate actions.

## 4. Arguments and default values

Let us extend `add_bonus()` so the caller can choose the bonus.
The new definition below replaces the earlier definition of `add_bonus`.

`score` is required. `bonus=2` provides a **default** that is used when the caller
does not supply a value for `bonus`.

````python
def add_bonus(score, bonus=2):
    new_score = score + bonus
    return new_score
````

````python
add_bonus(8)
````

**Positional arguments** fill the parameters in order: first `score`, then `bonus`.

````python
add_bonus(8, 5)
````

**Keyword arguments** identify a parameter by name. Mixing positional and keyword arguments
is allowed. In these calls, the positional argument comes first.

````python
add_bonus(8, bonus=5)
````

````python
add_bonus(bonus=5, score=8)
````

The last three calls all return the same value. Keyword arguments make the role of each value explicit.
The same idea applies to a built-in function: `sep` changes the separator used by `print`.

````python
print(1, 2, 3, sep = '-')
````

### Two useful argument errors

**Expected error — missing a required argument:** `score` has no default, so the caller must supply it.

````python
add_bonus()  # Expected TypeError: score is required.
````

**Expected error — supplying the same parameter twice:** positional `8` already fills `score`.
Adding `score=10` supplies it again. The problem is the duplicate value for `score`,
not the combination of positional and keyword arguments.

````python
add_bonus(8, score=10)  # Expected TypeError: score receives two values.
````

A valid mixed call supplies each parameter once:

````python
add_bonus(8, bonus=10)
````

## 5. Local names and outside names

A function's **parameters** and the names assigned in its body are normally **local** to that call.
In `add_bonus`, `score`, `bonus`, and `new_score` are local names.

Names assigned at the top level of this notebook are **global** names.
A local parameter can have the same spelling as a global name without changing that global binding.

````python
bonus = 100
print(add_bonus(8))
print(bonus)
````

The call uses the parameter's default, `bonus=2`. The outside name `bonus` still refers to `100`.

**Expected error:** the local name `new_score` is not defined at the top level.
Returning its value does not create a global name called `new_score`.

````python
new_score  # Expected NameError: this name is local to add_bonus.
````

### Rebinding a parameter is local

Compare this with Lecture 04's reassignment example. Assigning to a parameter name
changes that local binding; it does not reassign the caller's variable.

````python
x = 5
````

````python
def alter_x(x):
    x = x + 1
    return x
````

````python
alter_x(x)
````

````python
x
````

`alter_x(x)` returns `6`, while the outside `x` remains `5`.
To update the outside name, the caller can explicitly save the result.

````python
x = alter_x(x)
print(x)
````

Rebinding a name is different from changing a shared, mutable object.
A function can modify a list passed to it without using `global`;
we will examine that distinction when we study lists in Lecture 06.

### Reading an outside name

If a name is not local, Python can look it up in an enclosing function, the global scope,
or the built-in scope. Here `course_bonus` is global.
Reading it does **not** require the `global` keyword.

````python
course_bonus = 3

def use_outside_bonus(score):
    return score + course_bonus

use_outside_bonus(8)
````

This function depends on a value outside its parameter list.
Passing the bonus as an argument, as in `add_bonus`, makes the function's inputs explicit.

## 6. Apply the same ideas to strings

Now that we know how parameters and return values work, we can use a string method inside a function.
`phrase.upper()` produces an uppercase version of the string `phrase`.

````python
def shouting(phrase):
    shout = phrase.upper() + '!!!'
    return shout
````

````python
shouting('hi my name is miles')
````

**Expected error:** the integer `5` has no `.upper()` method.
This version of `shouting` expects a string.

````python
shouting(5)
````

If we want to accept values such as integers, we can deliberately convert the argument with `str()`.
The revised definition below makes that conversion before calling `.upper()`.

````python
def shouting(phrase):
    # attempt to convert the input object to a string
    shout = str(phrase).upper() + '!!!'
    return shout
````

````python
shouting(5)
````

````python
greeting = shouting("hi")
````

````python
greeting
````

### A final expression is not an implicit return

The notebook displays a cell's final expression, but a function body follows different rules.
Here `.lower()` produces a lowercase string, yet the function never returns it.
Predict the value assigned to `whisper`.

````python
def quiet(phrase):
    shh = str(phrase).lower()
    shh
````

````python
whisper = quiet("HELLO")
````

````python
print(whisper)
````

`quiet` returns `None`. To return the lowercase string, replace the final `shh` line with `return shh`.

### Follow a nested call

Arguments are evaluated before the outer function runs. This function prints its argument twice

````python
def print_twice(value):
    print(value)
    print(value)
````

## 7. A short extension: return several results

Python functions return one object. A comma-separated return expression packages several values
into a **tuple**. A tuple is an ordered collection; we will study it in more detail later.

````python
def powers_of(number):
    square = number ** 2
    cube = number ** 3
    return number, square, cube
````

````python
powers_of(3)
````

You can keep the tuple as one object or **unpack** its values into separate names.
For this unpacking form, the number of names must match the number of values.

````python
results = powers_of(3)
print(results)

number, square, cube = powers_of(3)
print(number, square, cube)
````

## 8. Where Python looks for names: LEGB

For the name lookups illustrated here, Python checks:

1. **Local:** names in the current function, including its parameters.
2. **Enclosing:** names in functions that lexically contain the current function.
3. **Global:** names at the top level of the module (this notebook, in our examples).
4. **Built-in:** names supplied by Python, such as `len`, `print`, and `ValueError`.

Keywords such as `def` and `return` are Python syntax, not names found through this lookup.
The examples below reset `x`, `y`, and `z` so we can compare their behavior directly.

### Scope Example 1: local and global names

````python
x, y, z = 1, 1, 1

def f():
    y = 2  # changing y to 2 only affects the value inside the function
    return x, y, z   # it does not find x or z in the local environment, so it searches the higher scope
````

````python
f() # the output of f is a tuple of the three values. 
# The value of y is 2, but x and z are still 1 because they were not changed in the local environment of f().
# When f returns x and z, it finds them in the global environment.
````

````python
print(x, y, z) # the values of x, y, and z in the global environment are unchanged by f().
````

The function uses its local `y = 2` and finds `x` and `z` globally.
Its local assignment leaves the global `y` unchanged.

### Scope Example 2: a function defined inside another function

````python
x, y, z = 1, 1, 1

def f():
    y = 2
    def g(): # we define a new function g inside f
        z = 3 # changing z to 3 only affects the value inside g, not f or the global environment
        return x, y, z 
    return g() 
````

````python
f() # when f is called, it calls g()
# When g() is called, it finds x in the global environment, y in the local environment of f(), and z in the local environment of g().
````

````python
print(x, y, z) # the values of x, y, and z in the global environment are unchanged by f() or g()
````

`g()` is defined inside `f()`. When `g` runs, it finds `z = 3` locally,
`y = 2` in the enclosing function `f`, and `x = 1` globally.
The call to `f()` returns `(1, 2, 3)`, while the global names remain `(1, 1, 1)`.

### Scope Example 3: definition location matters

````python
x, y, z = 1, 1, 1

def g(): # g is defined in the global environment
    z = 3
    return x, y, z

def f(): 
    y = 2
    return g()
````

````python
f() # when f() is called, it calls g(), which is defined in the global environment. 
# When g() is called, it finds x, y in the global environment. it does not use the value 2 from inside f.
````

````python
print(x, y, z)
````

Now `g()` is defined at the top level, outside `f()`.
It finds `x` and `y` globally, even though `f()` calls it.
The call to `f()` returns `(1, 1, 3)`.

The enclosing scopes come from **where a function is defined**, not from which function calls it.

## Optional appendix A: rebinding names with `global` and `nonlocal`

The main examples pass inputs as arguments and return results.
The following examples explore explicit rebinding outside a function.

- `global` makes assignments to the declared name bind at the module level.
- `nonlocal` makes assignments bind to an existing name in the nearest enclosing function scope.
- Neither declaration is needed merely to **read** an accessible outside name.

These examples are optional; they are not prerequisites for today's iClicker questions.

### Rebind a global name

Compare this with `alter_x`: `global x` makes the assignment in this function update
the global binding of `x`.

````python
def alter_global_x():
    global x
    x = x + 1
    return x
````

````python
x = 5
````

````python
alter_global_x()
````

````python
x
````

### Scope Example 4

````python
# global makes assignment to z update the global binding
x, y, z = 1, 1, 1

def f():
    y = 2
    def g():
        global z  # assignments to z in g use the global binding
        z = 3     # will assign 3 to the global variable z
        return x, y, z
    return g()
````

````python
f()
````

````python
print(x, y, z)
````

`g()` is defined inside `f()`. Its `global z` declaration makes `z = 3` update the global name.
For reads, `g` finds `y = 2` in `f` and `x = 1` globally.
The call returns `(1, 2, 3)`; afterward, global `x`, `y`, and `z` are `(1, 1, 3)`.

### Scope Example 5

````python
x, y, z = 1, 1, 1

def g():
    z = 3
    return x, y, z

def f():
    global y
    y = 2
    return g()
````

````python
g() # when we first run g(), it uses the global values of x and y, but the local value of z. Local value of z does not change global value z.
````

````python
print(x, y, z)
````

`g()` and `f()` are both defined in the global environment.

When `g()` runs, it sets `z = 3`. Inside `g()`, `x` and `y` are not defined. To find those values, it searches the higher scope which is the global environment because `g()` is defined inside the global environment. It uses the values in the global environment `x = 1` and `y = 1`.

````python
f()  # when we run f(), the global value of y is changed.
````

````python
print(x, y, z)
````

When we call the function `f()`, it modifies the value of `y` in the global environment. The final line of `f()` calls and returns the value of `g()`. This time, when `g()` looks for a value of `y`, it finds the value of `y` in the global environment which is now 2.

### Scope Example 6

````python
p, q = 1, 1

def f():
    global s   # the next assignment will bind s globally
    s = 2
    return p, q, s
````

````python
f()
````

````python
s
````

The `global` declaration alone does not create a value or binding.
The executed assignment `s = 2` creates or updates the global binding of `s`.

### Scope Example 7

````python
x, y, z = 1, 1, 1

def f():
    global y
    print("current value of y is " + str(y))
    y = 4
    def g():
        global y
        print("current value of y is now " + str(y))
        y = 10 
        print("current value of y is finally " + str(y))
        global z  
        z = 3
        return x, y, z
    return g()
````

````python
f()
````

````python
print(x, y, z)
````

Both `f()` and `g()` declare `global y`, so assignments to `y` in either function
update the same global binding. Reads of a global name do not require that declaration.

### Scope Example 8

````python
x, y, z = 1, 1, 1

def f():
    y = 4
    def g():
        nonlocal y  
        y = 10  # affects the y defined inside f
        global z  
        z = 3
        return x, y, z
    print(x, y, z)  # this line is run before g() is called
    return g()  # when g() is called, y will be modified
````

````python
f()
````

````python
print(x, y, z)
````

`f()` first creates a local `y = 4` and prints `(1, 4, 1)`.
Inside `g`, `nonlocal y` makes `y = 10` update that existing enclosing binding.
`global z` makes `z = 3` update the global binding.

The call returns `(1, 10, 3)`. The global `y` remains `1`, so the next print shows `(1, 1, 3)`.

### Scope Example 9: `nonlocal` requires an enclosing binding

**Expected error:** `f` has no enclosing function scope containing a binding for `r`.
Python rejects the definition with a `SyntaxError`; it does not run the function.

````python
p, q = 1, 1

def f():
    nonlocal r   # Expected SyntaxError: no binding for r in an enclosing function
    r = 2
    return p, q, r

f()
````

`nonlocal` requires an existing binding in an enclosing function scope.
A global binding does not satisfy this requirement, and `nonlocal` cannot create a new enclosing binding.

## Optional appendix C: read a traceback

A traceback shows the chain of active calls that led to an error.
Start with the error type and message at the bottom, then follow the frames to the failing line.
Exact cell numbers vary as you run the notebook.

In this deliberately broken example, `broken_print_twice` accepts a parameter named `value`,
but tries to read `joined`. That name is local to the calling function and is not in its enclosing scope.

````python
def broken_print_twice(value):
    print(joined)
    print(joined)

def combine_and_print(first, second):
    joined = first + " " + second
    broken_print_twice(joined)
````

````python
combine_and_print("bidi bidi", "bom bom")  # Expected NameError inside broken_print_twice.
````

Read the call chain: the notebook calls `combine_and_print`, which calls `broken_print_twice`,
which fails at `print(joined)`. The argument was passed successfully; the callee should use its parameter.

The working `print_twice(value)` defined earlier does that. Use it to repair the caller:

````python
def combine_and_print(first, second):
    joined = first + " " + second
    print_twice(joined)

combine_and_print("bidi bidi", "bom bom")
````
