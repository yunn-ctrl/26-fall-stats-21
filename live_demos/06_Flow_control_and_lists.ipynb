## 2. Choose a branch with `if`, `elif`, and `else`

An `if` / `elif` / `else` chain runs the block for the **first true condition**.
If every condition is false, the `else` block runs.

- End the `if` and `elif` lines with a colon.
- Indent the statements belonging to each block.
- Align `elif` and `else` with the original `if`.
- Parentheses around the condition are optional.

Try `x = -3`, `x = 0`, and `x = 3` in this example.

````python
x = -3

if x == 0:
    print(x, 'is zero')
elif x > 0:
    print(x, 'is positive')
else:
    print(x, 'must be negative')
````

### The order of the conditions matters

Both conditions below are true when `x = 100`, but only the first branch runs.
The `elif` condition is reached only if the earlier condition is false.

````python
x = 100

if x > 0:
    print(x, 'is positive')
elif x > 3:
    print(x, 'is greater than 3')  # Skipped because the earlier branch ran.
else:
    print(x, 'is zero or negative')
````

Two separate `if` statements perform two separate tests. Predict how this output differs.

## 3. Repeat a block with a `for` loop

A `for` loop runs its indented body once for each item in an **iterable**.
We will begin with `range`, an iterable of integers.

`range(4)` supplies `0`, `1`, `2`, and `3`. The stop value `4` is excluded.
On each iteration, `number` refers to the next value.

````python
for number in range(4):
    print(number)
print('The loop has finished')  # Outside the loop: this runs once.
````

### `range(start, stop, step)`

With one argument, `start` defaults to `0`; `step` defaults to `1`.
The stop value is always excluded. `range` arguments must be integers.

````python
for number in range(5, 10):
    print(number, end=' ')  # Use a space instead of a newline after each value.
````

````python
for number in range(0, 10, 2):
    print(number, end=' ')
````

````python
for number in range(3, 0, -1):
    print(number, end=' ')  # Negative step: count down toward the excluded stop.
````

### Accumulate a running total

Initialize the total **before** the loop. Update it inside the loop.
`total += number` is shorthand for `total = total + number` here.
Resetting `total` inside the loop would lose the earlier additions.

````python
total = 0
for number in range(1, 5):
    total += number
    print('Added:', number, 'Running total:', total)
print('Final total:', total)
````

A string is also iterable: the loop receives one character at a time.

````python
for letter in 'UCLA':
    print(letter)
````

### Encapsulate and generalize the working calculation

First get the calculation working with concrete values. Then put it into a function
and turn the changing input into a parameter. `sum_to(stop)` sums the integers from `1` through `stop`,
inclusive, for a nonnegative integer `stop`.

The loop stops before `stop + 1`. The `return` is outside the loop so it returns the completed total.

````python
def sum_to(stop):
    total = 0
    for number in range(1, stop + 1):
        total += number
    return total

print(sum_to(4))
print(sum_to(6))
````

## 4. Repeat while a condition is true

A `while` loop tests its condition **before each iteration**. It may run zero times.
For this countdown, `remaining -= 1` updates the value so the condition eventually becomes false.

````python
remaining = 3
while remaining > 0:
    print(remaining)
    remaining -= 1  # Shorthand for remaining = remaining - 1.
print('Blastoff!')
````

````python
remaining  # The loop stopped when remaining reached 0.
````

### Count upward instead of downward

Change the initial value, the condition, and the update together. The loop prints
`0` through `9`; the value is `10` when the next condition check ends the loop.

````python
i = 0
while i < 10:
    print(i, end=' ')
    i += 1
````

````python
i
````

### `break` ends the loop; `continue` skips to the next iteration

`break` exits the innermost loop. It does not merely skip the current item.
Here the loop encounters `5`, breaks, and never prints it or any later number.

````python
for number in range(1, 10):
    if number > 4:
        break
    print(number)
print('After the loop')
````

`continue` skips the rest of the current loop body and proceeds to the next iteration.
Here it skips printing each even number.

````python
for number in range(1, 7):
    if number % 2 == 0:
        continue
    print(number)
````

### The same `continue` pattern over a longer range

This range starts at `0`

````python
for n in range(20):
    # if the remainder of n / 2 is 0, skip the rest of the loop
    if n % 2 == 0:
        continue
    print(n, end=' ')
````

## Optional extension: recursion

Recursion is another way to express repetition: a function calls itself.
A recursive function needs a **base case**, and each recursive call must move toward it.
The base case alone is not enough if the argument never approaches it.

Compare this recursive countdown with the earlier `while` countdown.
For this example, supply a nonnegative integer `n`.

````python
def countdown(n):
    if n <= 0:
        print('Blastoff!')
    else:
        print(n)
        countdown(n - 1)
````

````python
countdown(3)
````

`countdown(3)` prints `3`, then calls `countdown(2)`. That prints `2` and calls `countdown(1)`.
The next call reaches `countdown(0)`, prints `Blastoff!`, and makes no further recursive call.
The waiting calls then finish, working back up the call chain.

### Trace the calls and their returns

- `countdown(3)` prints `3` and calls `countdown(2)`.
    - `countdown(2)` prints `2` and calls `countdown(1)`.
        - `countdown(1)` prints `1` and calls `countdown(0)`.
            - `countdown(0)` prints `Blastoff!` and returns without another call.
        - The call with `n = 1` finishes and returns.
    - The call with `n = 2` finishes and returns.
- The call with `n = 3` finishes and returns.

Starting directly at the base case makes no recursive call.

````python
countdown(0)
````

### Factorial has a natural recursive definition

For a nonnegative integer `n`, `n! = n * (n - 1)!`, with `0! = 1`.
This version rejects negative and non-integer inputs instead of silently returning an incorrect result.
`raise` explicitly reports an error; the rest of the definition implements the recurrence.

````python
def factorial(n):
    if type(n) != int:
        raise TypeError('n must be an integer')
    if n < 0:
        raise ValueError('n must be nonnegative')
    if n == 0:
        return 1
    return n * factorial(n - 1)

print(factorial(0))
print(factorial(4))
````

### Compare two factorial inputs

`4! = 4 * 3!`, `3! = 3 * 2!`, `2! = 2 * 1!`, `1! = 1 * 0!`, and `0! = 1`.
Changing the input from `4` to `5` adds one more multiplication and one more recursive call.

````python
factorial(4)
````

````python
factorial(5)
````

**Expected errors:** these inputs are outside the function's stated domain.

````python
factorial(-1)  # Expected ValueError.
````

````python
factorial(1.5)  # Expected TypeError.
````

### Another recursive printing example

For a nonnegative integer `n`, this function prints a string `n` times.
The bare `return` ends the base-case call and returns `None`, as in Lecture 05.

````python
def print_n(text, n):
    if n <= 0:
        return
    print(text)
    print_n(text, n - 1)

print_n('hello', 3)
````

The base-case input prints nothing. Compare it with the three repetitions above.

````python
print_n('hello', 0)
````
